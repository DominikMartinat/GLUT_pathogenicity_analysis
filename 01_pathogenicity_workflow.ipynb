### Identification of intracellular and extracellular termini
Analysis focuses on the intramembrane regions so we need to identify extramembrane head and tail of the sequence. Useful information will be lenghts of original protein and extramembrane termini and position where extramembrane tail starts in an original sequence.

```python
def get_extramembrane_parts(DTHMM_file):
    with open(DTHMM_file, "r") as f:
        lines = f.read().splitlines()
    annotation = lines[2].strip()

    whole_lenght = len(annotation)
    head_found = False

    for i in range(1, whole_lenght):

        if not head_found and annotation[i] == 'M':
            head_found = True
            head_length = i
        if annotation[i] != 'M' and annotation[i-1] == 'M':
            tail_length = whole_lenght - i
            new_end = i
    return {'head': head_length, 'tail': tail_length, 'length': whole_lenght, 'new_end': new_end}

    
```

```python
intramembrane_borders = {}
for p in proteins.values():
    intramembrane_borders[p] = get_extramembrane_parts(DTM_DIR / f'{p}.3line')
```

## Assign pathogenicites

Using AlphaMissense/PolyPhen-2/SIFT predictions and region files we generate new csv files equivalent to region files with added `pathogenicity` column. These are saved to `results/pathogenicity/{method dir}` directories.

```python
PAT_DIR = RES_DIR / 'pathogenicities'
AM_DIR = PAT_DIR / 'alphamissense'
PP_DIR = PAT_DIR / 'rhapsody'
SI_DIR = PAT_DIR / 'sift'

Path.mkdir(PAT_DIR, exist_ok=True)
Path.mkdir(AM_DIR, exist_ok=True)
Path.mkdir(PP_DIR, exist_ok=True)
Path.mkdir(SI_DIR, exist_ok=True)
```

First we generate files for complete proteins.

```python
for glut in proteins.keys():
    pm.alphamissense_residues(RES_DIR / f'pymissense/{proteins[glut]}-edit.pdb', AM_DIR / f'{reg_prefix["all"]}_{proteins[glut]}.csv')
    pm.polyphen2_residues(DATA_DIR / f'rhapsody/{glut}.txt', PP_DIR / f'{reg_prefix["all"]}_{proteins[glut]}.csv')
    pm.sift_residues(DATA_DIR / f'sift/SIFT_{glut}.txt', SI_DIR / f'{reg_prefix["all"]}_{proteins[glut]}.csv')  
```

Pathogenicity files for complete proteins are then used to assign pathogenicity to other regions in relation to their orientation.

Results for these pathogenicities can be found in `results/pathogenicities` directory.

```python
for up_id in proteins.values():
    am_patho_csv = AM_DIR / f'{reg_prefix["all"]}_{up_id}.csv'
    pp_patho_csv = PP_DIR / f'{reg_prefix["all"]}_{up_id}.csv'
    si_patho_csv = SI_DIR / f'{reg_prefix["all"]}_{up_id}.csv'

    #create file for rest of regions
    for reg in ['extracellular','transmembrane','intracellular']:
        analysis.assign_pathogenicity(am_patho_csv, region_csv = REG_DIR / f'{reg_prefix[reg]}_{up_id}.csv',out_file = AM_DIR / f'{reg_prefix[reg]}_{up_id}.csv')
        analysis.assign_pathogenicity(pp_patho_csv, region_csv = REG_DIR / f'{reg_prefix[reg]}_{up_id}.csv',out_file = PP_DIR / f'{reg_prefix[reg]}_{up_id}.csv')
        analysis.assign_pathogenicity(si_patho_csv, region_csv = REG_DIR / f'{reg_prefix[reg]}_{up_id}.csv',out_file = SI_DIR / f'{reg_prefix[reg]}_{up_id}.csv')
```

For further analysis we extract pathogenicities of priteins wthout hteir extramembrane termini.

These are then used for further analysis and can be found in `results/cut_pathogenicities` directory.

```python
#cutting put pathogenicities i am interested in
cut_PAT_DIR = RES_DIR / 'cut_pathogenicities'
Path.mkdir(cut_PAT_DIR, exist_ok=True)
#for all prediction methods
for pred in preds:
    #original preds
    pred_data_dir = RES_DIR / f'pathogenicities/{pred}'
    #new preds

    pred_res_dir = cut_PAT_DIR / pred
    Path.mkdir(pred_res_dir, exist_ok=True)


    #for all regions
    for reg in ['all', 'I', 'O', 'M']:

        #for all proteins - will not work for lining residues and binding pockets
        for prot in intramembrane_borders.keys():
            #read original predictions for region_protein
            df = pd.read_csv(pred_data_dir / f'{reg}_{prot}.csv')
            #throw away what is cut out
            df = df.query(f'{intramembrane_borders[prot]["head"]} < residue_label < {intramembrane_borders[prot]["new_end"]}')
            #renumber protein id
            df.loc[:,['residue_label']] = df.loc[:,['residue_label']] - intramembrane_borders[prot]['head']
            #save as csv
            df.to_csv(pred_res_dir / f'{reg}_{prot}.csv', index = False)
```

Pathogenicities of turncated proteins are then used to assign pathogenicities of lining residues and binding pockets.

Results can be found in `results/pathogenicities` direcotry.

```python

AM_DIR = cut_PAT_DIR / 'alphamissense'
PP_DIR = cut_PAT_DIR / 'rhapsody'
SI_DIR = cut_PAT_DIR / 'sift'



for up_id in proteins.values():
    am_patho_csv = AM_DIR / f'{reg_prefix["all"]}_{up_id}.csv'
    pp_patho_csv = PP_DIR / f'{reg_prefix["all"]}_{up_id}.csv'
    si_patho_csv = SI_DIR / f'{reg_prefix["all"]}_{up_id}.csv'

    #create file for rest of regions
    for reg in ['lining residues','binding pocket']:
        analysis.assign_pathogenicity(am_patho_csv, region_csv = REG_DIR / f'{reg_prefix[reg]}_{up_id}.csv',out_file = AM_DIR / f'{reg_prefix[reg]}_{up_id}.csv')
        analysis.assign_pathogenicity(pp_patho_csv, region_csv = REG_DIR / f'{reg_prefix[reg]}_{up_id}.csv',out_file = PP_DIR / f'{reg_prefix[reg]}_{up_id}.csv')
        analysis.assign_pathogenicity(si_patho_csv, region_csv = REG_DIR / f'{reg_prefix[reg]}_{up_id}.csv',out_file = SI_DIR / f'{reg_prefix[reg]}_{up_id}.csv')
```
